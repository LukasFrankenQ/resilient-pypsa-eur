```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

# Create qualitative plot showing electrification pathways
x = np.array([0, 1, 2])
x_labels = ['', 'Slower Electrification          ←          Expected Pathway          →          Faster Electrification', '']

# Create smooth curves using more points
x_smooth = np.linspace(0, 2, 100)

# First line: starts lower, ends higher (smooth curve with slight bend)
line1 = 0.4 + 0.08 * x_smooth - 0.05 * np.sin(x_smooth * np.pi / 3)

# Second line: starts higher, ends lower (smooth curve with opposite slope and slight bend)
line2 = 0.8 - 0.1 * x_smooth + 0.03 * np.sin(x_smooth * np.pi / 3)

# Plot the lines
ax.plot(x_smooth, line1, 'b-', linewidth=2, label='Lower bound')
ax.plot(x_smooth, line2, 'r-', linewidth=2, label='Upper bound')

# Fill the area between with red hatching
ax.fill_between(x_smooth, line1, line2, color='red', alpha=0.2, hatch='///', edgecolor='red')
ax.fill_between(x_smooth, line1, 0, color='b', alpha=0.2, edgecolor='b')

# Set x-axis ticks and labels
ax.set_xticks(x)
ax.set_xticklabels(x_labels)

# Add sample text above the red line
ax.text(1, 0.85, 'Cost during hikes in gas price', 
        ha='left', va='bottom', fontsize=10, style='italic', weight='bold')
ax.plot([0.7, 0.97], [0.76, 0.85], 'k-', linewidth=1)

ax.text(1.4, 0.3, 'Baseline cost', 
        ha='left', va='bottom', fontsize=10, style='italic', weight='bold')
ax.plot([1.1, 1.38], [0.43, 0.325], 'k-', linewidth=1)

# Remove y-axis for qualitative plot
ax.set_yticks([])

# Add grid for better readability
# ax.grid(True, alpha=0.3, axis='x')

# Add legend
# ax.legend()
ax.set_ylabel('Energy System Cost', labelpad=8)
ax.set_ylim(0.2, 1)

ax.set_xlabel('EU 2030s Energy Transition Progress', labelpad=8)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Add arrow heads to spines
ax.annotate('', xy=(0, 1.05), xytext=(0, 0.2),
            xycoords='data', textcoords='data',
            arrowprops=dict(arrowstyle='->', lw=1.5, color='black'),
            annotation_clip=False)

ax.annotate('', xy=(2.05, 0.2), xytext=(0., 0.2),
            xycoords='data', textcoords='data',
            arrowprops=dict(arrowstyle='->', lw=1.5, color='black'),
            annotation_clip=False)

ax.set_xlim(0, 2)
ax.set_xlim(0, 2)

plt.savefig('energy_system_cost_vs_electrification_progress.pdf')
plt.show()
```
